```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('SuperStoreOrders.csv')
```

```python
df.info()
```

```python
df.customer_name.value_counts().sort_values(ascending=False)
```

```python
top_selling_item = df.product_name.value_counts().sort_values(ascending=False).head(10)
```

```python
top_selling_item
```

```python
plt.bar(top_selling_item.index, height=top_selling_item)
plt.title('Top Selling Items')
plt.xlabel('Items')
plt.ylabel('Quantity')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
top_sale_country = df.country.value_counts().sort_values(ascending=False).head(10)
```

```python
plt.bar(top_sale_country.index, height=top_sale_country)
plt.title('Top Sales Country')
plt.xlabel('Country')
plt.ylabel('Quantity')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
product_category = df.category.value_counts().sort_values(ascending=False)
```

```python
plt.pie(product_category, labels=product_category.index, autopct='%1.1f%%')
plt.title('Product Category Distribution')
plt.show()
```

```python
product_sub_category = df.sub_category.value_counts().sort_values(ascending=False).head(10)
```

```python
plt.pie(product_sub_category, labels=product_sub_category.index, autopct='%1.1f%%')
plt.title('Product Sub Category Distribution')
plt.show()
```

```python
# Select product name and category which generate the highest revenue
highest_revenue_product = df.groupby(['product_name', 'category'])['sales'].sum().sort_values(ascending=False).head(10)
```

```python
plt.bar(highest_revenue_product.index.get_level_values('product_name'), height=highest_revenue_product.values)
plt.title('Highest Revenue Product')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
